### Question 1 

The price,  $P(\lambda)$ of a five-year bond is given by the formula

$$ P(\lambda) = \frac{7\lambda^4 +35\lambda^3 +70\lambda^2 +70\lambda + 235}{2(1+\lambda)^5} $$

where $\lambda$ is the effective annual yield

4. The gradient of the price-yield curve for this bond at 𝜆 = 10% is −316.1316. Find, to 2 decimal places, the duration of the bond at 𝜆 = 10%.

#### i) Find, to 2 decimal places, the price of the bond when 𝜆 = 10%

Just substitute

##### Solution
$\begin{align} P(0.1) &= \frac{7\times0.1^4 +35\times0.1^3 +70\times0.1^2 +70\times0.1 + 235}{2(1+0.1)^5} \\ \notag &= 75.36  \end{align}$

#### ii) Given that this bond pays coupons annually at a rate of c, is redeemable at par and has a face value of 100, determine the value of c. 

##### Core Principle
Since annual payment $m=1, n=t =5$ from part i we have $P(\lambda) = 75.36$ and $F=100$

$$ P(\lambda) = \frac{F}{(1+\lambda)^n} + Fc\frac{1-(1+\lambda)^{-n}}{\lambda} $$

##### Solution
$\begin{align} 75.36 &= \frac{100}{(1+0.1)^5} + 100c\frac{1-(1+0.1)^{-5}}{0.1} \end{align}$

$\begin{align} c &= \left(75.36 - \frac{100}{(1+0.1)^5}\right) \div \left(100\frac{1-(1+0.1)^{-5}}{0.1}\right) \\ \notag &= 0.035\end{align}$

#### iv) The gradient of the price-yield curve for this bond at 𝜆 = 10% is −316.1316.  Find,  to 2 decimal places, the duration of the bond at 𝜆 = 10%.


##### Core Principle
$$D(\lambda) =(1+\lambda) D_{mod}(\lambda) =(1+\lambda)\times \left(-\frac{P'(\lambda)}{P(\lambda)}\right) $$

So why?

Definition of (Macaulays) duration $D(\lambda)$: present-value-weighted average time to the bond's cash flows

$$D(\lambda) = \frac{\sum_{i=1}^{n} \frac{iCF_i}{(1+\lambda)^i}} {\sum_{i=1}^{n} \frac{CF_i}{(1+\lambda)^i}} = \frac{\sum_{i=1}^{n} \frac{iCF_i}{(1+\lambda)^i}} {P(\lambda)} $$

Definition of modified duration $D_{mod}(\lambda)$

$$D_{mod}(\lambda) = -\frac{P'(\lambda)}{P(\lambda)} $$

Find $P'(\lambda)$

$$\begin{align} P'(\lambda) &= \frac{d}{d\lambda} \left( \frac{Fc}{1+\lambda} + \frac{Fc}{(1+\lambda)^2} +... \frac{F(c+1)}{(1+\lambda)^n}  \right) \\ \notag &= - \left( \frac{1Fc}{(1+\lambda)^2} + \frac{2Fc}{(1+\lambda)^3} +... \frac{nF(c+1)}{(1+\lambda)^{n+1}}  \right) \\ \notag &= -\frac{1}{(1+\lambda)} \sum_{i=1}^{n}\frac{iCF_i}{(1+\lambda)^i} \end{align}$$

Hence the relationship between Duration and Modified duration is

$$\begin{align} D_{mod}(\lambda) &= -\frac{P'(\lambda)}{P(\lambda)} \\ \notag &=   -\frac{-\frac{1}{(1+\lambda)} \sum_{i=1}^{n}\frac{iCF_i}{(1+\lambda)^i}} {P(\lambda)} \\ \notag &= \frac{1}{(1+\lambda)} \frac{ \sum_{i=1}^{n}\frac{iCF_i}{(1+\lambda)^i}} {P(\lambda)} \\ \notag &=  \frac{1}{(1+\lambda)} D(\lambda) \end{align}$$

##### Solution

$D(0.1) =(1+0.1) D_{mod}(0.1) =(1.1)\times \left(-\frac{-316.1316}{75.36}\right) = 4.61$

### Question 2 
A 10-year bond pays annual coupons of 5%. It is redeemable at par and is priced at par. 
Calculate 

1. the modified duration w.r.t. the effective annual yield of the bond;
2. the modified duration w.r.t. the semiannually compounded yield of the bond;
3. the modified duration w.r.t. the continuously compounded yield of the bond;

#### i) the modified duration w.r.t. the effective annual yield of the bond

##### Core Principle
Apply same principle as Question 1 iv, note it is priced at par so $\lambda = c = 0.05$

$$D_{mod}(\lambda) = -\frac{P'(\lambda)}{P(\lambda)}$$

##### Solution

$P\begin{align}(\lambda) =  \frac{F}{(1+\lambda)^n} + Fc\frac{1-(1+\lambda)^{-n}}{\lambda}  \end{align}$

$\begin{align} P'(\lambda) &=  -\frac{nF}{(1+\lambda)^{n+1}} - Fc\left(\frac{1-(1+\lambda)^{-n}}{\lambda^2} - \frac{n(1+\lambda)^{-(n+1)}}{\lambda} \right) \\ \notag &= -\frac{nF}{(1+\lambda)^{n+1}} - F\frac{1-(1+\lambda)^{-n}}{\lambda} + \frac{nF}{(1+\lambda)^{n+1}} \\ \notag &= - F\frac{1-(1+\lambda)^{-n}}{\lambda} \end{align}$  

(note $c = \lambda$)

$\begin{align} P'(0.05) &= - F\frac{1-(1+0.05)^{-10}}{0.05} = -7.7217F\end{align} $  
$\begin{align} P(0.05) &=  F \end{align}$  

$\begin{align}D_{mod}(0.05) = -\frac{-7.7217F}{F} = 7.7217\end{align}$

```python
(1-1.05**-10)/0.05
```

#### ii) the modified duration w.r.t. the semiannually compounded yield of the bond; 

##### Core Principle
**COMMON MISTAKE**: Assuming a change in compounding frequency not changing the rate

$$D_{mod}(\lambda^{(m)}) = -\frac{P'(\lambda^{(m)})}{P(\lambda^{(m)})}$$

$\lambda^{(2)} = 2\left((1+\lambda)^{0.5} -1\right) = 2\left(\sqrt{1.05} -1\right) = 0.04939$

hence note

$\left((1+\lambda^{(2)}/2)^{2} -1\right) = \lambda = c$

##### Solution

$\begin{align}P((1+\lambda^{(2)}/2)^{2} -1) = \frac{F}{(1+\lambda^{(2)}/2)^{2n}} + Fc\frac{1-(1+\lambda^{(2)}/2)^{-2n}}{(1+\lambda^{(2)}/2)^{2} -1}  \end{align}$

Tedious calculation below

$\begin{align} (P((1+\lambda^{(2)}/2)^{2} -1))' &=  -\frac{nF}{(1+\lambda^{(2)}/2)^{2n+1}} - Fc\left(\frac{\left(1-(1+\lambda^{(2)}/2)^{-2n}\right) (1+\lambda^{(2)}/2) }{\left((1+\lambda^{(2)}/2)^{2}-1\right)^2} - \frac{n(1+\lambda^{(2)}/2)^{-(2n+1)}}{(1+\lambda^{(2)}/2)^{2} -1} \right) \\ \notag &= -\frac{nF}{(1+\lambda^{(2)}/2)^{2n+1}} - Fc\frac{\left(1-(1+\lambda^{(2)}/2)^{-2n}\right) (1+\lambda^{(2)}/2) }{\left((1+\lambda^{(2)}/2)^{2}-1\right)^2} + \frac{nFc}{(1+\lambda^{(2)}/2)^{2n+1}c} \\ \notag &= - Fc\frac{1-(1+\lambda^{(2)}/2)^{-2n}}{\left((1+\lambda^{(2)}/2)^{2}-1\right)^2}(1+\lambda^{(2)}/2) \\ \notag &= - Fc\frac{1-(1+\lambda)^{-n}}{\lambda^{2}}(1+\lambda^{(2)}/2) \\ \notag &= - F\frac{1-(1+\lambda)^{-n}}{\lambda}(1+\lambda^{(2)}/2) \\ \notag &=-FD_{mod}(\lambda)\sqrt{1+\lambda} \end{align}$  


Resulting equation

$\begin{align} (P((1+\lambda^{(2)}/2)^{2} -1))' &= -FD_{mod}(\lambda)\sqrt{1+\lambda} = -7.9124F\end{align}$  

or rather 

$\begin{align} (P((1+\lambda^{(2)}/2)^{2} -1))' &= \frac{-FD(\lambda)}{\sqrt{1+\lambda}} = -7.9124F\end{align}$  

Final Answer

$D_{mod}(\lambda^{(2)}) = 7.9124 $



##### Another way to simplify it

$\begin{align}
(P((1+\lambda^{(2)}/2)^{2} -1))'
&=\frac{d}{d\lambda^{(2)}} P'(\lambda) \\
&= P'(\lambda)\sqrt{1+\lambda}.
\end{align}$

$\begin{align}
D_{mod}(\lambda^{(2)}) &= \frac{P'(\lambda^{(2)})}{P(\lambda^{(2)})}\\
&= \frac{P'(\lambda)\sqrt{1+\lambda}}{P(\lambda)}\\
&= D_{mod}(\lambda)\sqrt{1+\lambda}
\end{align}$

Generalized

$\begin{align}
D_{mod}(\lambda^{(n)}) &= \frac{P'(\lambda^{(n)})}{P(\lambda^{(n)})}\\
&= \frac{P'(\lambda)(1+\lambda)^{1-1/n}}{P(\lambda)}\\
&= D_{mod}(\lambda)(1+\lambda)^{1-1/n}
\end{align}$

#### iii) the modified duration w.r.t. the continuously compounded yield of the bond; and iv) the Macaulay duration

##### Core Principle
Interesting case lets try

$\lambda^{(\infty)} = log(1.05) = 0.04879$

Also recall the identity

$lim_{k\to \infty} (1+\lambda^{(k)}/k)^{k} = exp\left\{ \lambda^{(\infty)}\right\} =  exp\left\{ log(1.05)\right\} = 1.05$

##### Solution

$\begin{align} lim_{k\to \infty} P(\lambda^{(k)}) = lim_{k\to \infty}\left(\frac{F}{(1+\lambda^{(k)}/k)^{kn}} + Fc\frac{1-(1+\lambda^{(k)}/k)^{-kn}}{(1+\lambda^{(k)}/k)^{k} -1} \right) \end{align}$

Tedious calculation below

$\begin{align} lim_{k\to \infty} P'(\lambda^{(k)}) &=  lim_{k\to \infty} \left(-\frac{nF}{(1+\lambda^{(k)}/k)^{kn+1}} - Fc\left(\frac{\left(1-(1+\lambda^{(k)}/k)^{-kn}\right) (1+\lambda^{(k)}/k )^{k-1}}{\left((1+\lambda^{(k)}/k)^{k}-1\right)^2} - \frac{n(1+\lambda^{(k)}/k)^{-(kn+1)}}{(1+\lambda^{(k)}/k)^{k} -1} \right) \right) \\ \notag &= lim_{k\to \infty} \left(-\frac{nF}{(1+\lambda^{(k)}/k)^{kn+1}} + \frac{nFc}{(1+\lambda^{(k)}/k)^{kn+1}c} \right) - lim_{k\to \infty} \left(Fc\frac{\left(1-(1+\lambda^{(k)}/k)^{-kn}\right) (1+\lambda^{(k)}/k )^{k-1}}{\left((1+\lambda^{(k)}/k)^{k}-1\right)^2}\right) \\ \notag &=  -lim_{k\to \infty} \left(Fc\frac{1-(1+\lambda^{(k)}/k)^{-kn}}{\left((1+\lambda^{(k)}/k)^{k}-1\right)^2}(1+\lambda^{(k)}/k )^{k-1}\right) \\ \notag &= - Fc\frac{1- exp\left\{ -n\lambda^{(\infty)}\right\}}{ \left(exp\left\{ \lambda^{(\infty)} \right\}-1\right)^2} exp\left\{ \lambda^{(\infty)}\right\} \\ \notag&= - Fc\frac{1-(1+\lambda)^{-n}}{\lambda^2}(1+\lambda) \\ \notag &= - F\frac{1-(1+\lambda)^{-n}}{\lambda}(1+\lambda) \\ \notag &=-FD_{mod}(\lambda)(1+\lambda) \end{align}$  


Resulting equation

$\begin{align} lim_{k\to \infty} P'(\lambda^{(k)}) &= -FD_{mod}(\lambda)(1+\lambda) = -8.1078F\end{align}$  

or rather 

$\begin{align} lim_{k\to \infty} P'(\lambda^{(k)}) &= -FD(\lambda) = -8.1078F\end{align}$  

Final Answer

$D_{mod}(\lambda^{(\infty)}) = D(\lambda) =  8.1078 $


### Question 3 
A n-year bond and a 2n-year bond, both paying coupons annually, have the same annual yield to maturity, $\lambda$. It is also given that the coupon rate is equal to the yield rate. If the Macaulay duration of the n-year bond is k times the Macaulay duration of the 2n-year bond for some $k\in(0.5,1)$

Express $\lambda$ in terms of k and n

#### Core principle
Write the formula down, note again $c = \lambda$, $P(\lambda) = F$

$\begin{align}D(\lambda) =D_{mod}(\lambda)(1+\lambda) = -\frac{P'(\lambda)}{P(\lambda)}(1+\lambda)\end{align}$

1. For maturity n

$\begin{align}P_n(\lambda) =  \frac{F}{(1+\lambda)^n} + Fc\frac{1-(1+\lambda)^{-n}}{\lambda}  \end{align}$

$\begin{align} D_n(\lambda) &= \left(\frac{1-(1+\lambda)^{-n}}{\lambda}\right) (1+\lambda)  \end{align}$  
$~$

2. For maturity 2n

$\begin{align}P_{2n}(\lambda) =  \frac{F}{(1+\lambda)^{2n}} + Fc\frac{1-(1+\lambda)^{-2n}}{\lambda}  \end{align}$

$\begin{align} D_{2n}(\lambda) &= \left(\frac{1-(1+\lambda)^{-2n}}{\lambda}\right) (1+\lambda)  \end{align}$  

##### Solution

$\begin{align} kD_{2n}(\lambda) &= D_n(\lambda)\end{align}$  


$\begin{align} k\left(\frac{1-(1+\lambda)^{-2n}}{\lambda}\right) (1+\lambda) &= \left(\frac{1-(1+\lambda)^{-n}}{\lambda}\right) (1+\lambda)\end{align}$ 

$\begin{align} k &=\frac{1-(1+\lambda)^{-n}}{1-(1+\lambda)^{-2n}} \end{align}$

#### iii) Given that the convexity at 𝜆 = 0.04 is 1081.08. Find the approximate percentage change in bond price when the bond yield decreases by 10 basis points from 4%.


##### Core principle
First understand how to find the change in bond price $\Delta P$ given modified duration $D_{mod}$, Convexity $C$, current price $P$ and change in yield $\Delta \lambda$

$\Delta P(\lambda) \approx P'(\lambda)\Delta \lambda + \frac{1}{2}P''(\lambda)(\Delta \lambda)^2$

where

$D_{mod} = -\frac{P'(\lambda)}{P(\lambda)}$

$C = \frac{P''(\lambda)}{P(\lambda)}$


##### Solution

$\begin{align} \frac{\Delta P(\lambda)}{P(\lambda)} &\approx \frac{P'(\lambda)}{P(\lambda)}\Delta \lambda + \frac{1}{2}\frac{P''(\lambda)}{P(\lambda)}(\Delta \lambda)^2 \\ \notag &= -D_{mod}\Delta \lambda  +\frac{C}{2}(\Delta \lambda)^2 \\ \notag &\approx 2.5125 \% \end{align}$
