### Question 1 
An investor with initial wealth $\frac{25}{9}$ and utility function $U(w) = \sqrt{w} ~~, w>0 $ following lottery:

| Outcome | Probability |
|:---------------:|:--------------:|
| Lose 1 | $3p$ |
| Win $\frac{8}{3}$ | $p$ |
| Win $\frac{11}{9}$ |$1-4p$ |

Find the range of 𝑝 for which the investor will avoid the lottery.

##### Core principle
Recall the expected utility of the lottery follows

$$E(U(w_0+x)) = \sum_{i=1}^{n}p(x_i)U(w_0+x_i)$$

Find 

$$E(U(w_0+x)) < U(w_0)$$

##### Solution

$\begin{align} E(U(w_0+x)) &= 3p \sqrt{\frac{25}{9}-1} + p \sqrt{\frac{25}{9}+\frac{8}{3}} + (1-4p)\sqrt{\frac{25}{9}+\frac{11}{9}}  
\\ &= 3p \frac{4}{3} +p \frac{7}{3}+ 2(1-4p)
\\ &= 2 -\frac{5}{3}p < \sqrt{\frac{25}{9}}
\end{align}$

Hence

$\begin{align} p >\frac{1}{5} \end{align}$

notice **outcome 3** has probability distribution $1-4p$, hence $p \leq \frac{1}{4}$

Answer: $\begin{align} \frac{1}{5} <p \leq \frac{1}{4} \end{align}$

### Question 2

#### i) Find the greatest possible value that x can take (so that 𝐶(𝑥) to be defined).

##### Core principle
For $C(x)$ to be define, the possible values of $x$ must ensure $U(w_0+x)$ is real since $U(w) = \sqrt{w}$

$$w_0+x\geq0$$

##### Solution

$\begin{align} 12+2x \geq 0

$\begin{align}0<x\leq 8\end{align}$


#### ii) Find the value of x that maximizes 𝐶(𝑥)

##### Core principle

$$ C(x) = U^{-1} E(U(w_0+x)))$$

\end{align}$

$\begin{align} C(x) &= U^{-1} E(U(w_0+x)))

\end{align}$

Hence

$\begin{align} C'(x) &= \frac{d}{dx}\left[ 6+ \frac{x}{8} + \frac{1}{2}\sqrt{144+6x-3x^2} \right]

### Question 3

#### i) Find in terms of r and s the certainty equivalent of this investment. 

##### Core principle
Define $C(x)$ 

$$ C(x) = U^{-1} E(U(w_0+x)))$$

##### Solution

$\begin{align} E(U(w_0+x))) &= 0.8 ln\left(20000\cdot (1+\frac{r}{100})\right)+  0.2 ln\left(20000\cdot (1-\frac{s}{100})\right)

\\  &=  ln\left(20000\cdot (1+\frac{r}{100})^{\frac{4}{5}}(1-\frac{s}{100})^{\frac{1}{5}}\right)
\end{align}$

$\begin{align} U^{-1}(x) &= e^{x}
\end{align}$

Hence

$\begin{align}C(x) &= U^{-1} E(U(w_0+x)))
\\ &= e^{ln\left(20000\cdot (1+\frac{r}{100})^{\frac{4}{5}}(1-\frac{s}{100})^{\frac{1}{5}}\right)}
\\ &= 20000\cdot (1+\frac{r}{100})^{\frac{4}{5}}(1-\frac{s}{100})^{\frac{1}{5}}

### Question 4
Investor A has utility function $U_A(w) = 𝑤^\alpha$ , where $w,\alpha > 0$.

#### i) Find the range of 𝛼 such that investor A is risk averse. 

##### Core principle
A risk averse investor has a concave Utility Function where the 2nd derivative is negative

$$  U''_A(w) < 0$$

##### Solution

$\begin{align} U''_A(w) &= \frac{d^2}{dw^2}w^\alpha =\alpha (\alpha-1) w^{\alpha -2} <0
\end{align}$

Since $\alpha, w >0$ 

$0<\alpha<1$
