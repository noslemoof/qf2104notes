### Question 1 

The present value, P of a perpetual annuity that pays $A$ at $t = 1,2, 3 \dots$ is given by 

$$P=\frac{A}{\lambda}$$

Derive the Macaulay’s duraion, D of this perpetual annuity in two ways. 


#### (i) Use the equation $\frac{dP/d\lambda}{P}= -D_M$ to find $D_M$ and hence find D.


##### Core principle
Just differentiate using the formula given

##### Solution

$\frac{dP}{d\lambda}= -\frac{A}{\lambda^2}$

$P=\frac{A}{\lambda}$

$\begin{align} D_M = -\frac{-\frac{A}{\lambda^2}}{\frac{A}{\lambda}} \end{align} = \frac{1}{\lambda}$

$\begin{align}D = D_M (1+\lambda) =\frac{1+\lambda}{\lambda}  \end{align}$

#### (ii) Find D directly from the definition $D =\begin{align}\frac{\sum_{i=1}^{\infty}iA(1+\lambda)^{-i}}{\sum_{i=1}^{\infty}A(1+\lambda)^{-i}}\end{align}$

hint: $\begin{align}\sum_{i=1}^{\infty}iy^{i}= \frac{y}{(1-y)^2}\end{align}$ 

##### Core principle
Use the hint appropriately, notice 

$$\begin{align}\sum_{i=1}^{\infty}iy^{-i}= \sum_{i=1}^{\infty}i(1/y)^{i}=\frac{1/y}{(1-1/y)^2}\end{align}$$

Also notice

$$\begin{align}P = \sum_{i=1}^{\infty}A(1+\lambda)^{-i}\end{align}$$

##### Solution
let $y = (1+\lambda)$

$\begin{align} D =\frac{\sum_{i=1}^{\infty}iA(1+\lambda)^{-i}}{\sum_{i=1}^{\infty}A(1+\lambda)^{-i}} = \frac{A\frac{1/y}{(1-1/y)^2}}{A/(y-1)} = \frac{\frac{(1+\lambda)^{-1}}{(1-(1+\lambda)^{-1})^2}}{\lambda^{-1}} = \lambda\frac{(1+\lambda)}{\lambda^2} = \frac{1+\lambda}{\lambda} \end{align}$

###  Question 2
Let $N > 100$ be a positive integer. In terms of the integer $N$, find the duration of the following cash flow $(1, 0, 2, 0, 3, 0, 4\dots, N)$, given that the yield is zero for the first 2N years. Give your answer in simplified exact term. 

Hint: $\begin{align}\sum_{i=1}^{N} i  = \frac{N(N+1)}{2}\end{align} $, $\begin{align}\sum_{i=1}^{N}i^2 = \frac{N(N+1)(2N+1)}{6} \end{align} $

##### Core principle
Algebraic manipulation: identify patterns and formulate simply



Definition of duration: 

$$\begin{align}D =\frac{\sum_{i=1}^{N}iCF_i(1+\lambda)^{-i}}{\sum_{i=1}^{N}CF_i(1+\lambda)^{-i}}\end{align}$$

But since $\lambda=0,$ $(1+\lambda)^{-i} = 1, ~~~ \forall i$

$$D =\begin{align}\frac{\sum_{i=1}^{N}iCF_i}{\sum_{i=1}^{N}CF_i}\end{align}$$

Simply put **we multipy the ith year to the corresponding cashflow**

Perspective 1: consider cash flow at ith order and find the corresponding year

Perspective 2: consider the ith year and find the corresponding cash flow

##### Solution
**Perspective 1:**
1. Cashflow 1: $CF_1 = 1$, year = 0
2. Cashflow 2: $CF_2 = 2$, year = 2
3. Cashflow 3: $CF_3 = 3$, year = 4
4. Cashflow N: $CF_N = N$, year = 2N-2

Hence for the numerator: 

$\begin{align}\sum_{i=1}^{N}iCF_i &= \sum_{i=1}^{N}(2i-2)\cdot i \\ \notag &= 2\sum_{i=1}^{N}i^2-2\sum_{i=1}^{N}i \end{align}$

and the denominator: 

$\begin{align}\sum_{i=1}^{N}CF_i = \sum_{i=1}^{N} i \end{align} $

Solving them:

$\begin{align} 2\sum_{i=1}^{N}i^2-2\sum_{i=1}^{N}i = \frac{N(N+1)(2N+1)}{3} - N(N+1) \end{align}$

$\begin{align}\sum_{i=1}^{N} i  = \frac{N(N+1)}{2}\end{align}$

$\begin{align}D  &= \frac{\frac{N(N+1)(2N+1)}{3} - N(N+1)}{\frac{N(N+1)}{2}} \\ \notag &= \frac{2}{3}\left(2N+1-3 \right) = \frac{4}{3}(N-1) \end{align}$

**Perspective 2:**
1. Year 0: $CF_0 = 1$
2. Year 1: $CF_1 = 0$ 
3. Year 2: $CF_2 = 2$
4. Year 3: $CF_3 = 0$
5. Year 2N-2: $CF_{2N-2} = N$

Since all odd years have cashflow 0, we will consider every 2 years starting from year 0 ending at $N-1$ for every 2 interval, and the cash flow for the corresponding ith year is $i+1$

Hence for the numerator: 

$\begin{align}\sum_{i=0}^{N-1}2iCF_i &= \sum_{i=0}^{N-1}2i (i+1)  \\ \notag &= 2\sum_{i=0}^{N-1}i^2+2\sum_{i=0}^{N-1}i \end{align}$

and the denominator: 

$\begin{align}\sum_{i=0}^{N-1}CF_i = \sum_{i=0}^{N-1} i+1 \end{align} $

Solving them:

$\begin{align} 2\sum_{i=0}^{N-1}i^2+2\sum_{i=0}^{N-1}i &= \frac{(N-1)(N)(2N-1)}{3} + N(N-1) = \frac{2}{3}N(N-1)(N+1)\end{align}$

note the 0th term adds nothing so its essentially equivalent to adding from the first term :)

$\begin{align}\sum_{i=0}^{N-1} i+1  = \frac{N(N-1)}{2} + N = \frac{N(N+1)}{2} \end{align}$

$\begin{align}D  &= \frac{\frac{2}{3}N(N-1)(N+1)}{\frac{1}{2}N(N+1)} \\ \notag &= \frac{4}{3}(N-1) \end{align}$

### Question 4 
Let $U(w) = ln (w)$ be the utility function of an investor whose initial wealth is $2\alpha, ~(\alpha> 1)$. The 
investor is offered an investment that pays $2\alpha$ and $-\alpha$ with probabilities 1/3 and 2/3 
respectively. 

#### i) Prove analytically that the investor will not invest. 

##### Core principle
Recall

$E(U(w_0+x)) = \sum_{i=1}^{n}p(x_i)U(w_0+x_i)$

Remember to mention $p(x_i)$ is the probability of payout $x_i$

##### Solution

$\begin{align}E(U(\alpha+x)) &= p(2\alpha) U(4\alpha) + p(-\alpha) U(\alpha) \\ \notag &= \frac{1}{3} ln (4\alpha) + \frac{2}{3} ln(\alpha) \\ \notag &= \frac{1}{3} ln(4\alpha^3) = ln(4^{1/3}\alpha) < ln(2\alpha) = U(w_0)\end{align}$

Since $E(U(\alpha+x)) < U(w_0)$, he will not invest

#### ii) Illustrate this result graphically.

i will draw it assuming $\alpha=10$

Note our outcome concidentally is below the the Initial welath

```python
import matplotlib.pyplot as plt
import numpy as np

alpha = 10

# Wealth outcomes
w_bad = alpha          # 10
w_good = 4 * alpha     # 40

# Probabilities
p_bad = 2/3
p_good = 1/3

# Expected wealth and expected utility
EW = p_bad * w_bad + p_good * w_good
EU = p_bad * np.log(w_bad) + p_good * np.log(w_good)

# Utility curve
space = np.linspace(5, 50, 500)
utility = np.log(space)

plt.plot(space, utility, label=r"$U(w)=\ln(w)$")

# Bad and good outcome points
plt.scatter(w_bad, np.log(w_bad), marker="x", s=100,
            label="Bad outcome")
plt.scatter(w_good, np.log(w_good), marker="x", s=100,
            label="Good outcome")

# Straight line joining bad and good outcomes
plt.plot(
    [w_bad, w_good],
    [np.log(w_bad), np.log(w_good)],
    linestyle="--",
    label="Line between outcomes"
)

# Expected outcome point
plt.scatter(EW, EU, marker="o", s=100,
            label=r"$(E[W],E[U(W)])$")

# Show U(E[W]) on utility curve
plt.scatter(EW, np.log(EW), marker="x", s=100,
            label=r"$U(E[W])$")

# Vertical line at expected wealth
plt.axvline(EW, linestyle="dotted")

# Horizontal expected utility line
plt.axhline(EU, linestyle="dotted",
            label=r"$E[U(W)]$")

plt.xlabel("Wealth")
plt.ylabel("Utility")
plt.title("Expected Utility and Risk Aversion")

plt.legend()
plt.show()
```

What if we shift the probability distribution 

say if we shift to $p_{win} = 5/6$, or $p_{lose} = 1/6$

```python
import matplotlib.pyplot as plt
import numpy as np

alpha = 10

# Wealth outcomes
w_bad = alpha
w_good = 4 * alpha
w_initial = 2 * alpha

space = np.linspace(5, 50, 500)
utility = np.log(space)


def plot_gamble(p_win):

    p_lose = 1 - p_win

    # Expected wealth
    EW = p_win * w_good + p_lose * w_bad

    # Expected utility
    EU = (
        p_win * np.log(w_good)
        + p_lose * np.log(w_bad)
    )

    # Utility curve
    plt.figure(figsize=(8, 5))
    plt.plot(space, utility, label=r"$U(w)=\ln(w)$")

    # Bad and good outcomes
    plt.scatter(w_bad, np.log(w_bad), marker="x", s=100,
                label="Bad outcome")

    plt.scatter(w_good, np.log(w_good), marker="x", s=100,
                label="Good outcome")

    # Chord between outcomes
    plt.plot(
        [w_bad, w_good],
        [np.log(w_bad), np.log(w_good)],
        linestyle="--",
        label="Line between outcomes"
    )

    # Expected utility point
    plt.scatter(EW, EU, s=100,
                label=r"$(E[W], E[U(W)])$")

    # Initial wealth
    plt.scatter(
        w_initial,
        np.log(w_initial),
        marker="x",
        s=100,
        label=r"$U(2\alpha)$"
    )

    # Expected wealth
    plt.axvline(
        EW,
        linestyle="dotted",
        label=r"$E[W]$"
    )

    # Expected utility
    plt.axhline(
        EU,
        linestyle="dotted",
        label=r"$E[U(W)]$"
    )

    # Utility of not playing
    plt.axhline(
        np.log(w_initial),
        linestyle="--",
        label=r"$U(2\alpha)$"
    )

    plt.xlabel("Wealth")
    plt.ylabel("Utility")
    plt.title(
        f"Probability of Winning = {p_win:.2f}"
    )

    # Same axes for both graphs
    plt.xlim(5, 50)
    plt.ylim(2.1, 3.9)

    plt.legend()
    plt.show()


# Plot 1: low probability of winning
plot_gamble(1/6)

# Plot 2: high probability of winning
plot_gamble(5/6)
```
