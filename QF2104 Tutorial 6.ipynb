### Question 1 
You are given two risky assets, A and B whose rates of returns are given in the table below:

| Stock | Mean | Variance |
|:---------------:|:--------------:|:--------------:|
| A | 0 | 0.1 | 
| B | 0.2 | 0.15 |

It is also given that the covariance of the return rates is −0.1. 

#### i) Find the equation of the feasible set (that is, the $\mu$-$\sigma$ curve). 

##### Core principle
Recall the following formula for portfolio return and variance, where the $\alpha$ is the proportion invested in stock A

$$\mu_p = \alpha\mu_A+(1-\alpha)\mu_B$$

$$\sigma_p^2 = \alpha^2\sigma_A^2+(1-\alpha)^2\sigma_B^2 + 2\alpha(1-\alpha)\sigma_{AB}$$

Define $\alpha$ in terms of $\mu_p,\mu_A,\mu_B$ and define $\sigma_p$ in terms of the aforementioned and $\sigma_A,\sigma_B,\sigma_{AB}$

##### Solution
Substituting values for $\mu_A=0, \mu_B=0.2$

$\begin{align}
&\mu_p = 0.2(1-\alpha)\\
&5\mu_p = (1-\alpha)\\
&\alpha= 1-5\mu_p\\
\end{align}$

Substituting values for $\sigma_A^2=0.1, \sigma_B^2=0.15$

$\begin{align} \sigma_p^2 
&= 0.1 \alpha^2+0.15(1-\alpha)^2- 0.2\alpha(1-\alpha)\\
&=0.1 (1-5\mu_p)^2+0.15(5\mu_p)^2 - (1-5\mu_p)\mu_p \\
&=0.1- \mu_p + 2.5\mu_p^2+3.75\mu_p^2-\mu_p+5\mu_p^2\\
&=\frac{45}{2}\mu_p^2-2\mu_p+\frac{1}{10}\\
\end{align}$

#### (ii) Calculate the portfolio weights that will minimize the portfolio risk if 
(a) there are no restrictions \
(b) the desired mean portfolio return is at least 8% \
(c) the desired mean portfolio return is at least 9%.

##### Core principle
Re-use the following equations for a,b,c. Determine the first derivative of $\sigma$ with respect to $\alpha$, for a) find the critical point with respect to $\alpha$ and notice the value of mean for the corresponding $\alpha$

$$\mu_p = \alpha\mu_A+(1-\alpha)\mu_B$$

$$\sigma_p^2 = \alpha^2\sigma_A^2+(1-\alpha)^2\sigma_B^2 + 2\alpha(1-\alpha)\sigma_{AB}$$

##### Solution
a)

$\begin{align}
\frac{\delta \sigma_p^2}{\delta \alpha} 
&= 2\alpha\sigma_A^2-2(1-\alpha)\sigma_B^2 + 2(1-2\alpha)\sigma_{AB}\\
&= 0.2\alpha-0.3(1-\alpha) - 0.2(1-2\alpha)\\
&= 0.9\alpha -0.5 = 0
\end{align}$

$\alpha = \frac{5}{9},\quad (1-\alpha) =\frac{4}{9} $

b) Consider from a) that $\alpha=\frac{5}{9}$, find $\mu_p$ for that $\alpha$

$$\mu_p = 0.2(1-\frac{5}{9})\approx 0.0889$$

Hence $\alpha = \frac{5}{9},\quad (1-\alpha) =\frac{4}{9} $, since the mean at min variance portfolio is already greater than 0.08

c)

Also consider $\frac{\delta \mu_p}{\delta \alpha} = -0.2$. this means $\mu$ is a decreasing function with respect to $\alpha$. When finding the min variance with a fixed minimal mean $\mu_p=0.09$, we choose $\alpha_{\mu_p = 0.09}$.

$\begin{align}
&0.09 = 0.2(1-\alpha)\\
&\alpha=0.55\\
&1-\alpha = 0.45\\
\end{align}$


### Question 2

coefficient given by 

$$\mu_A=0.2,\quad \mu_B=0.3,\quad \sigma_A=0.1,\quad \sigma_B=0.2,\quad \rho_{AB}=\rho$$


#### i) Find the range of values of 𝜌 for which the global minimum-variance portfolio does not involve short-selling. 

##### Core principle
Use the following formula

$$\sigma_p^2 = \alpha^2\sigma_A^2+(1-\alpha)^2\sigma_B^2 + 2\alpha(1-\alpha)\sigma_{AB}$$

##### Solution
Substituting values for $\sigma_A=0.1, \sigma_B=0.2$

$\begin{align}
\frac{\delta \sigma_p^2}{\delta \alpha} 
&= 2\alpha\sigma_A^2-2(1-\alpha)\sigma_B^2 + 2\sigma_A\sigma_B(1-2\alpha)\rho_{AB}\\
&= 0.02\alpha-0.08(1-\alpha) + 0.04\rho(1-2\alpha)\\
&= (0.1-0.08\rho)\alpha -(0.08-0.04\rho) = 0
\end{align}$

Hence

$$\alpha = \frac{0.08-0.04\rho}{0.1-0.08\rho}=\frac{1}{2}+\frac{0.03}{0.1-0.08\rho}$$

Fnally find $0\leq\alpha\leq1$

$\begin{align} 
&0\leq\frac{1}{2}+\frac{0.03}{0.1-0.08\rho}\leq1\\
&-0.5\leq\frac{0.03}{0.1-0.08\rho}\leq0.5\\
&-0.5(0.1-0.08\rho)\leq 0.03 \leq 0.5(0.1-0.08\rho)\\
&-0.05+0.04\rho\leq 0.03 \leq 0.05-0.04\rho\\
&\rho\leq2,\quad \rho\leq0.5\\
&-1\leq\rho\leq0.5
\end{align}$

### Question 3 
A portfolio is to be constructed from two assets whose mean and variance of return rate 
are summarized in the table below. 

| Asset | Mean | Variance |
|:---------------:|:--------------:|:--------------:|
| 1 | 0 | 0.2 | 
| 2 | 0.2 | 0.8 |

The correlation of the return rates of these assets is $\rho$ if $\rho=\frac{5}{8}$

#### i) Find the minimum-risk portfolio and the corresponding mean and variance 

##### Core principle
Use the following formula again

$$\sigma_p^2 = \alpha^2\sigma_1^2+(1-\alpha)^2\sigma_2^2 + 2\alpha(1-\alpha)\rho_{12}\sigma_1\sigma_2$$
$$\mu_p = \alpha\mu_1+(1-\alpha)\mu_2$$
##### Solution
$\begin{align} 
\sigma_p^2&= 0.2\alpha^2+ 0.8(1-\alpha)^2+0.5\alpha(1-\alpha)\\
&=0.2\alpha^2+0.8-1.6\alpha+0.8\alpha^2+0.5\alpha-0.5\alpha^2\\
&=0.5\alpha^2-1.1\alpha+0.8\\
&\frac{\delta \sigma_p^2}{\delta \alpha} = \alpha-1.1=0\\
&\alpha = 1.1\\
&1-\alpha = -0.1
\end{align}$

Thus we find $\mu_p = \alpha\mu_1+(1-\alpha)\mu_2 = -0.02$ and $\sigma_p^2 = 0.195 $

#### ii) Sketch the $(\sigma_p,\mu_p)$ graph vary. Indicate the points representing the two assets

##### Core principle
Use part 1 solutions

$$\sigma_p^2 = 0.5\alpha^2-1.1\alpha+0.8$$
$$\alpha=1-5\mu_p$$

##### Solution
$\begin{align} 
\sigma_p^2&= 0.5\alpha^2-1.1\alpha+0.8\\
&= 0.5(1-5\mu_p)^2-1.1(1-5\mu_p)+0.8\\
&=0.5 -5\mu_p +12.5\mu_p^2 -1.1+5.5\mu_p +0.8\\
&= 12.5\mu_p^2 + 0.5\mu_p +0.2\\
\end{align}$

```python
import numpy as np
import matplotlib.pyplot as plt
a1 = (-0.02,0.195**0.5)
s1 = (0,0.2**0.5)
s2 = (0.2,0.8**0.5)
alpha = np.linspace(-0.5,2.5,3000)
mu = (1-alpha)*0.2
sigma2 = 0.5*alpha**2-1.1*alpha+0.8
sigma = np.sqrt(sigma2)
# Asymptotes
x = np.linspace(0, max(sigma), 3000)
plt.figure(figsize=(10,10))
plt.plot(x, -0.02 + x/np.sqrt(12.5), '--', label="Upper asymptote")
plt.plot(x, -0.02 - x/np.sqrt(12.5), '--', label="Lower asymptote")


plt.plot(sigma,mu,label = "mu = f(sigma)")
plt.axhline(0,color="black")
plt.axhline(-0.02,color="green",linestyle="-.")
plt.scatter(a1[1],a1[0],marker='x',label=f"GMVP:({a1[0]},{a1[1]:.3f})")
plt.scatter(s1[1],s1[0],marker='x',label=f"stock 1:({s1[0]},{s1[1]:.3f})")
plt.scatter(s2[1],s2[0],marker='x',label=f"stock 2:({s2[0]},{s2[1]:.3f})")
plt.xlim(left=0)
plt.legend()
```

$$\frac{\delta \sigma_p^2}{\delta \alpha} = \alpha-1.1$$

Notice $\sigma_p^2$ is an decreasing function with respect to $\alpha$ when $0\leq\alpha\leq1$, hence we choose $\alpha=1$

##### Solution

Thus $\mu_p = =\mu_1 = 0$ and $\sigma_p= \sqrt{0.2}$

#### iii) Find the weight, mean and standard deviation of the portfolio that has the smallest variance when the portfolio mean is at least 0.1 

##### Core principle
Consider part iii) where we see

$$\sigma_p^2 = 0.5\alpha^2-1.1\alpha+0.8$$
$$\mu_p = 0.2(1-\alpha)$$

$$\frac{\delta \sigma_p^2}{\delta \alpha} = \alpha-1.1$$

$\sigma_p^2$ is an decreasing function with respect to $\alpha$ and $\frac{\delta \mu}{\delta \alpha} = -0.2<0$ also a decreasing function. Hence notice we can simply acheive the minimum portfolio when $\mu_p=0.1$

##### Solution

Thus $\sigma_p^2 = 12.5\mu_p^2 + 0.5\mu_p +0.2 = 0.375$, and also $\alpha = 0.5$

#### ii) If $\rho = −1$, sketch the $(\sigma_p,\mu_p)$-graph, and calculate the mean and variance of the minimum-risk portfolio.

##### Core principle
Redo i,ii,iii with new $rho=-1$

##### Solution
$\begin{align}

&=1.8\alpha^2-2.4\alpha+0.8\\
&\frac{\delta \sigma_p^2}{\delta \alpha} = 3.6\alpha-2.4=0\\

Thus we find $\mu_p = 1/15$ and $\sigma_p^2 = 0$. $\alpha = 2/3$ and $1-\alpha = 1/3$

$\begin{align}
