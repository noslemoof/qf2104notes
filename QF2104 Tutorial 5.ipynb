### Question 1
Consider two utility functions, $U(x)$ and $V(x)=1-\frac{1}{U(x)}$, where $U(x)>0$ for all real x. The Arrow-Pratt absolute risk aversion function corresponding to $U$ and $V$ are denoted respectively by $R_U$ and $R_V$. It is given that $U$ is strictly concave. 

#### i) By considering $V''$, prove $V$ is strictly concave

##### Core principle
Find and prove $V'' < 0$. For concave $V$
##### Solution

$\begin{align} V' &= \frac{U'}{U^2}
\end{align}$

$\begin{align} V'' &= \left(\frac{U'}{U^2}\right)'
\\ &= \frac{U''}{U^2} - \frac{2(U')^2}{U^3} < 0
\end{align}$

Since 

$\begin{align} U'' < 0, \quad (U')^2 > 0, \quad U>0 \end{align}$


#### ii) Show
$$R_V = R_U+\frac{2U'}{U}$$
and thus conclude that an investor with utility function 𝑉 is globally more risk 
averse than an investor with utility function 𝑈. 

##### Core principle
$$R_V = -\frac{V''}{V'}$$
##### Solution

$\begin{align} R_V &= -\frac{U''/U^2 - 2(U')^2/U^3}{U'/U^2}
\\ &= -U''/U' + 2 U'/U 
\\ & = R_U +\frac{2U'}{U}
\\ & >R_U
\end{align}$

Since 

$\begin{align} U' > 0, \quad U>0 \end{align}$


### Question 2

It is given that $U(x)>0, \forall x\in \mathbb{R}$, and $V(w) = 1-e^{-U(w)}$. Suppose Both A and B have initial wealth $w_0$ and are offered a lottery. Let $C_A$ and $C_B$ be the certainty equivalent of the lottery for A and B respectively

#### i) By expressing $R_V$ in terms of $R_U$, prove that investor B is globally more risk averse than investor A. 

##### Core principle
Find $V'$ and $V''$ first and derive
##### Solution

$\begin{align} V' &= U'e^{-U}
\end{align}$

$\begin{align} V'' &= (U'e^{-U})'
\\ &= U''e^{-U} - (U')^2e^{-U}
\end{align}$

$\begin{align} R_V &= - \frac{ U''e^{-U} - (U')^2e^{-U}}{ U'e^{-U}}
\\ &=-U''/U' +U'
\\ &= R_U + U' 
\\ &> R_U
\end{align}$

Since $ U' > 0$. Hence $R_V > R_U, \forall w$

\end{align}$

$$ f(k) = E(U(w_0+rk))$$

$$ f(k) = E(U(w_0-k+k(1+r))) $$

Find value of $f(k)$ at critical points

### Question 4
Another measure of risk aversion known as the Relative Risk Aversion (RRA) coefficient, is defined by

$$U_{RRA}=-w\cdot\frac{U''(w)}{U'(w)}=(wU_{ARA}),\quad w>0$$

For a concave function $U$

#### a) Show that each of the following utility function has a constant RRA.
i) $U(w) = w^\alpha, 0<\alpha<1$

ii) $U(w) = ln~ w$

##### Core principle
Simply plug in the result of your 1st, 2nd derivative and w

##### Solution i)

$\begin{align} U(w) &= w^{\alpha},
\\ U'(w) &= \alpha w^{\alpha-1} > 0,
\\ U''(w) &= \alpha(\alpha-1) w^{\alpha-2} < 0, \quad \alpha-1<0
\end{align}$

$\begin{align} U_{RRA}(w) &= -w\cdot \frac{\alpha(\alpha-1) w^{\alpha-2}}{ \alpha w^{\alpha-1}}
\\ &= 1-\alpha
\end{align}$

##### Solution ii)

$\begin{align} U(w) &= ln~w,
\\ U'(w) &= \frac{1}{w} > 0,
\\ U''(w) &= -\frac{1}{w^2} < 0
\end{align}$

$\begin{align} U_{RRA}(w) &= -w\cdot \frac{-w^{-2}}{w^{-1}}
\\ &= 1
\end{align}$


#### b) The hyperbolic absolute risk aversion (HARA) class of utility functions is defined by
$$U(w)=\frac{1-\lambda}{\lambda} \left(\frac{aw}{1-\lambda}+b\right)^\lambda +c,\quad w>0$$
where $a,b,c$ and $\lambda>0$ are parameters

i) Find the ARA and RRA for this function in terms of its parameters

ii) By setting $a=b=1,c=0$, and taking limit $\lambda \to\infty$ show that the utility function $U(w) = -e^{-w}$ is a member of the HARA class (in the limiting sense). (recall $\underset{t\to\infty}{\text{lim}}(1+x/t)^t=e^x$)

##### Core principle
Work out the derivatives and take careful steps not to miss params through calculations

##### Solution i)
$\begin{align} U(w) &= \frac{1-\lambda}{\lambda} \left(\frac{aw}{1-\lambda}+b\right)^\lambda +c,
\\ U'(w) &= \frac{1-\lambda}{\lambda}\frac{a\lambda}{1-\lambda} \left(\frac{aw}{1-\lambda}+b\right)^{\lambda-1} 
\\ &=  a\left(\frac{aw}{1-\lambda}+b\right)^{\lambda-1}
\\ U''(w) &= -a^2\left(\frac{aw}{1-\lambda}+b\right)^{\lambda-2}
\end{align}$

$\textbf{ARA}$

$\begin{align} R_U &=-\frac{U''(w)}{U'(w)}
\\  &= -\frac{-a^2\left(\frac{aw}{1-\lambda}+b\right)^{\lambda-2}}{a\left(\frac{aw}{1-\lambda}+b\right)^{\lambda-1}}
\\ &=  a\left(\frac{aw}{1-\lambda}+b\right)^{-1}
\end{align}$

$\textbf{RRA}$

$\begin{align} U_{RRA}(w) &= wU_{ARA}
\\ &= aw\left(\frac{aw}{1-\lambda}+b\right)^{-1}
\end{align}$

##### Solution ii)

$\begin{align} \underset{\lambda\to\infty}{\text{lim}}U(w) &=  \underset{\lambda\to\infty}{\text{lim}}
\frac{1-\lambda}{\lambda} \left(\frac{w}{1-\lambda}+1\right)^\lambda \quad \text{Let}~t=\lambda-1
\\ &= \underset{t\to\infty}{\text{lim}}-\frac{t}{t+1} \left(1+\frac{-w}{t}\right)^{t+1}
\\  &= \underset{t\to\infty}{\text{lim}}-\left(1-\frac{1}{t+1}\right) \left(1+\frac{-w}{t}\right)^{t}\left(1+\frac{-w}{t}\right)
\\ &= -(1)\cdot e^{-w}\cdot(1) 
\\ &= -e^{-w}
\end{align}$
