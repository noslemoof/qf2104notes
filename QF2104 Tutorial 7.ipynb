### Question 1
Consider a financial market of n risky assets. Use the Lagrangian Method to find the weight 
vector of Global Minimum-Variance portfolio. 

#### Core Principles
Recall the formulas for finding the portfolio Variance with $w$, $C$ and its constraints

$$\sigma_p^2 = w^T Cw = \sum_{j=1}^{n} \sum_{i=1}^{n} w_jw_i\sigma_{ij}$$
$$\sum_{i=1}^{n}w_i = w^T\textbf{1}=1$$
$$L(w_1,w_2,\dots w_n,\lambda) = \frac{1}{2} \sigma_p^2 -\lambda\left(\sum_{i=1}^nw_i-1\right)$$

Thus differentiate with respect to $w_i$ and $\lambda$

#### Solution

\begin{flalign}
&\frac{\partial L}{\partial w_i} = Cw -\lambda \textbf{1} = 0 &&\\
&\frac{\partial L}{\partial \lambda} =  1 - w^T\textbf{1} = 0 &&\\
\end{flalign}

For the first partial derivative we have

\begin{flalign}
&Cw =\lambda \textbf{1}&&\\
& w =\lambda C^{-1}\textbf{1}  &&\\
\end{flalign}

Then for the second partial derivative we have

\begin{flalign}
&w^T\textbf{1} = 1 &&\\
&(\lambda C^{-1}\textbf{1})^T\textbf{1} = \lambda\textbf{1}^TC^{-1} \textbf{1}=1 &&\\
&\frac{1}{\lambda} = \textbf{1}^TC^{-1} \textbf{1} &&\\
&\lambda = \frac{1}{\textbf{1}^TC^{-1} \textbf{1}} &&\\
\end{flalign}

Using the first Partial derivative again we have

$$w^* = \frac{C^{-1}\textbf{1}}{\textbf{1}^TC^{-1} \textbf{1}}$$

It is given that $C^{-1} = \begin{pmatrix}25&5&-50\\p&5&-10\\-50&q&r\end{pmatrix}$, find p q and r

##### Core Principles
Assume that the covariance matrix $\textbf{C}$ is positive definite. That is $\forall x\ne0$

$$x^TCx>0$$

Thus no nonzero linear combination of asset returns can have 0 variance in our case. This is necessary for use to find an inverse matrix. Properties of a Positive Definite Covariance matrix

1. All Eigenvalues are strictly positive
2. Det(C) >0
3. $C^{-1}$ exists
4. The Function $w^TCw$ is stricly convex

These conditions enforce a unique GMVP

#### ii)
Given further that $C^{-1}\mu = \begin{pmatrix}0\\2\\2\end{pmatrix}$, find the equation $\sigma^2 = A\mu^2 +B\mu+C$ pf the minimum-variance frontier for the 3 risky assets

##### Core Principles
For practice and learning sake let use build our minimum variance frontier from first principles. We define our variance as follows

$$\sigma_p^2 = w^T C w$$

Subject to the constraints

$$w^T \textbf{1} = 1$$

$$ w^T\mu-\mu_p =0 $$

We then define our lagrangian

$$L(w_1,w_2,\dots w_n,\lambda_1,\lambda_2) = \frac{1}{2} w^T C w -\lambda_1\left(w^T \textbf{1} -1\right) -\lambda_2\left(w^T\mu-\mu_p\right)$$

By differentiation against $w, \lambda_1, \lambda_2$ we find  

\begin{flalign}
&\frac{\partial L}{\partial w_i} = Cw -\lambda_1 \textbf{1}- \lambda_2\mu = 0 &&\\
&\frac{\partial L}{\partial \lambda_1} =  1 - w^T\textbf{1} = 0 &&\\
&\frac{\partial L}{\partial \lambda_2} =  \mu_p-w^T\mu  = 0 &&\\
\end{flalign}

For the first partial derivative we have

\begin{flalign}
& w =\lambda_1 C^{-1}\textbf{1}-\lambda_2C^{-1}\mu &&\\
\end{flalign}

Then for the second partial derivative we have

\begin{flalign}
&\textbf{1}^T w = 1 &&\\
&\textbf{1}^T\left(\lambda_1 C^{-1}\textbf{1}-\lambda_2C^{-1}\mu\right) =1 &&\\
&\lambda_1 \textbf{1}^TC^{-1}\textbf{1}-\lambda_2\textbf{1}^TC^{-1}\mu =1 &&\\
\end{flalign}

\begin{flalign}
&\mu^T w = \mu_p &&\\
&\mu^T\left(\lambda_1 C^{-1}\textbf{1}-\lambda_2C^{-1}\mu\right) =\mu_p &&\\
&\lambda_1 \mu^TC^{-1}\textbf{1}-\lambda_2\mu^TC^{-1}\mu = \mu_p &&\\
\end{flalign}

let $a = \textbf{1}^TC^{-1}\textbf{1},\quad b =\mu^TC^{-1}\textbf{1}= \textbf{1}^TC^{-1}\mu, \quad c= \mu^TC^{-1}\mu $

Hence, solve the simultaneous equation

$$1 = \lambda_1 a+\lambda_2 b$$
$$\mu_p = \lambda_1 b+\lambda_2 c$$


\begin{flalign}
&b - a\mu_p = \lambda_2 (b^2 - ac)&&\\
&\lambda_2 = \frac{a\mu_p-b }{ac- b^2}&&\\
&\lambda_1 = \frac{c-b\mu_p}{ac- b^2}&&\\
\end{flalign}

Finally substitute our first partial derivative
\begin{flalign}
&Cw -\lambda_1 \textbf{1}- \lambda_2\mu = 0 &&\\
&Cw -\frac{c-b\mu_p}{ac- b^2} \textbf{1}- \frac{a\mu_p-b }{ac- b^2}\mu = 0 &&\\
&w =C^{-1}\left(\frac{c-b\mu_p}{ac- b^2} \textbf{1}+ \frac{a\mu_p-b }{ac- b^2}\mu\right)&&\\
\end{flalign}

Our minimum variance frontier thus follows
\begin{flalign}
\sigma_p^2 = w^TCw &= w^TC C^{-1}\left(\frac{c-b\mu_p}{ac- b^2} \textbf{1}+ \frac{a\mu_p-b }{ac- b^2}\mu\right)&&\\
&= w^T\left(\frac{c-b\mu_p}{ac- b^2} \textbf{1}+ \frac{a\mu_p-b }{ac- b^2}\mu\right)&&\\
&= \frac{c-b\mu_p}{ac- b^2} w^T\textbf{1}+ \frac{a\mu_p-b }{ac- b^2}w^T\mu&&\\
&= \frac{c-b\mu_p}{ac- b^2} + \frac{a\mu_p-b }{ac- b^2}\mu_p&&\\
&= \frac{a\mu_p^2-2b\mu_p+c}{ac-b^2}
\end{flalign}

#### Solutions
\begin{flalign}
&a = \textbf{1}^TC^{-1}\textbf{1} = \begin{pmatrix}1&1&1\end{pmatrix}\begin{pmatrix}25&5&-50\\5&5&-10\\-50&-10&120\end{pmatrix}\begin{pmatrix}1\\1\\1\end{pmatrix} =40&&\\
&b =\textbf{1}^TC^{-1}\mu= \begin{pmatrix}1&1&1\end{pmatrix}\begin{pmatrix}0\\2\\2\end{pmatrix} =4&&\\
&c= \mu^TC^{-1}\mu= \begin{pmatrix}0.1&0.5&0.1\end{pmatrix}\begin{pmatrix}0\\2\\2\end{pmatrix} =\frac{6}{5}&&\\
\end{flalign}

\begin{flalign}
&A =\frac{a}{ac-b^2} = \frac{40}{48-16} =\frac{5}{4}&&\\
&B =\frac{-2b}{ac-b^2} = \frac{-8}{48-16} =-\frac{1}{4}&&\\
&C =\frac{c}{ac-b^2} = \frac{6/5}{48-16} =\frac{3}{80}&&\\
\end{flalign}

$$\sigma_p^2 = \frac{5}{4}\mu_p^2 - \frac{1}{4}\mu_p +\frac{3}{80}$$

#### iii) Find the mean and variance of the global minimum-variance portfolio.

##### Core Principles
Our GMVP lies on the minimum variance frontier formula. Hence just differentiate with respect to $\mu_p$
$$\sigma_p^2 = \frac{5}{4}\mu_p^2 - \frac{1}{4}\mu_p +\frac{3}{80}$$

#### Solutions
$$\frac{\partial \sigma_p^2}{\partial \mu_p} = \frac{5}{2}\mu_p - \frac{1}{4} =0 $$

$$\mu_p = \frac{1}{10}$$
$$\sigma_p^2 = \frac{1}{40}$$

### Question 3
Let $w_{fund1} =\frac{C^{-1}\textbf{1}}{\textbf{1}^TC^{-1}\textbf{1}}$ and $w_{fund2} =\frac{C^{-1}\mu}{\textbf{1}^TC^{-1}\mu}$ and $a = \textbf{1}^TC^{-1}\textbf{1}$, $b=\textbf{1}^TC^{-1}\mu>0$ and $c = \mu^TC^{-1}\mu$ a portfolio, $p$ has a rate of return $r_p$ and portfolio mean $\mu_p$. Another portfolio whose weight vector is $\alpha w_{fund1}+ (1-\alpha)w_{fund2}$ has a rate of return $r_q$. It is given that $\mu_p>\frac{b}{a}$ and $q$ is uncorrelated to $p$. Thus $Cov(r_p,r_q) = 0$

#### i) Find $\alpha$ in terms of $a,b$ and $\mu$.
[Hint: Covariance between portfolio p and q is $w_p^TCw_q$]

##### Core Principles
Since we know
$$Cov(r_p,r_q) = w_p^TCw_q $$

Use the provided information $w_q = \alpha w_{Fund1} + (1-\alpha) w_{Fund2}$

##### Solutions
\begin{flalign}
w_p^TCw_q  &=w_p^TC(\alpha w_{Fund1} + (1-\alpha) w_{Fund2})&&\\
&=\alpha w_p^TC\frac{C^{-1}\textbf{1}}{\textbf{1}^TC^{-1}\textbf{1}}+ (1-\alpha)w_p^TC\frac{C^{-1}\mu}{\textbf{1}^TC^{-1}\mu}&&\\
&=\alpha \frac{w_p^T\textbf{1}}{a}+ (1-\alpha)\frac{w_p^T\mu}{b}&&\\
&=\alpha \frac{1}{a}+ (1-\alpha)\frac{\mu_p}{b} = 0 &&\\
\end{flalign}

$$\alpha = \frac{\mu_p/b}{\mu_p/b - 1/a}= \frac{a\mu_p}{a\mu_p - b}$$

#### ii) Determine if q is an efficient portfolio

##### Core Principles
Notice the returns for fund 1 and fund 2

$$\mu_1 = \mu^T\frac{C^{-1}\textbf{1}}{\textbf{1}^TC^{-1}\textbf{1}} = \frac{b}{a}$$
$$\mu_2 = \mu^T\frac{C^{-1}\mu}{\textbf{1}^TC^{-1}\mu} = \frac{c}{b}$$

Can u determine the returns of $\mu_q$?

##### Solutions
\begin{flalign}
\mu_q  &= \alpha \mu_1 + (1-\alpha) \mu_2&&\\
&=\frac{b}{a}\frac{a\mu_p}{a\mu_p - b} + \frac{c}{b} (1-\frac{a\mu_p}{a\mu_p - b})&&\\
&=\frac{b\mu_p - c}{a\mu_p-b}&&\\
\end{flalign}

Check if returns $\mu_q>\mu_1$ since $\mu_1$ corresponds to the GMVP

$$\frac{b\mu_p - c}{a\mu_p-b} - \frac{b}{a} = \frac{ac - b^2}{a(b-a\mu_p)}<0$$ 

Notice we take $C$ as a positive definite matrix. Hence $ac-b^2 >0$. Since $\mu_p>\frac{b}{a}$ we have $a(b-a\mu_p)<0$

Since $\mu_q <\mu_1$ its returns are less than the GMVP. Implying it is NOT efficient
